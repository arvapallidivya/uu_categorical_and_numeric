```python
# Code that can be used for Question 1

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

# Number of possible values in our grid
K = 20

# Number of times we repeat the experiment
n = 5000

# Create the grid:
# 1/K, 2/K, ..., K/K
grid = np.arange(1, K + 1) / K

# Randomly choose n values from the grid.
# Each value has the same probability of being selected.
draws = np.random.choice(grid, size=n)

# Plot the ECDF of our simulated values
sns.ecdfplot(x=draws)

# Add the theoretical Uniform(0, 1) CDF
x = np.linspace(0, 1, 500)
plt.plot(x, stats.uniform.cdf(x, loc=0, scale=1))

plt.xlabel("Value")
plt.ylabel("ECDF")
plt.title("Question 1: ECDF of Simulated Values")

plt.show()

# %%
#Official Code for Question 1
np.random.seed(2)
n=5000
K = [5, 50, 100, 1000, 2000, 50000]
for k in K:
    grid = np.arange(1, k + 1) / k # discrete uniform distribution
    draws = np.random.choice(grid, size=n) # simulating 5000 times
    x = np.sort(draws)
    y = np.arange(1, n + 1) / n
    print(f"\n\n================= Current K = {k} ==================================\n")
    plt.figure(figsize=(4,4))
    plt.step(x, y, where="post")
    plt.xlabel("x")
    plt.ylabel("ECDF")
    plt.title("Empirical CDF")
    plt.show()
```

```python
#Question 3  - n values changing
N = [3, 10, 40, 100, 1000]
for n in N:
    p = 0.5         # probability of heads, held fixed
    b = 5000        # number of simulations

    flips = np.random.random((b, n)) < p        # b experiments, n flips each
    number_of_heads = np.sum(flips, axis=1)      # heads count per experiment

    expected_heads = n * p # theoretical mean (number of flips * probability)
    standard_deviation = np.sqrt(n * p * (1 - p))

    std_heads = (number_of_heads - expected_heads) / standard_deviation

    print(f"\n\n================= Current n = {n} ==================================\n")
    print(np.mean(std_heads))

    # plot ecdf
    x = np.sort(std_heads)
    y = np.arange(1, b + 1) / b
    plt.figure(figsize=(5,4))
    plt.step(x, y, where="post", label="Simulated ECDF")

    # theoretical standard normal for comparison
    x_theory = np.linspace(-4, 4, 500)
    plt.plot(x_theory, stats.norm.cdf(x_theory, loc=0, scale=1), linestyle="--", label="Standard Normal")

    plt.xlabel("Standardized number of heads")
    plt.ylabel("ECDF")
    plt.title(f"ECDF of Standardized Coin Flips (n={n})")
    plt.legend()
    plt.grid(alpha=0.3)
    plt.show()

```

```python
# Base code on Q4

# Question 4

r = 1.0 # average rate at which the event arrives, as r increases our probability increases, r is the rate of something happening so if the rate is higher, so if 2 buses come per minute
dt = 1e-3 # time interval -> like seconds
n = 5000

# Probability of termination during each tiny time interval
probability = r * dt

# Make sure the probability is valid
if probability >= 1:
    raise ValueError("r * dt must be less than 1.")

# Store the waiting times
waiting_times = []

# Repeat the simulation 5,000 times
for i in range(n):

    # Start at period 0
    periods = 0

    # Keep going until termination happens
    while True:

        # Move forward one time period
        periods += 1

        # Generate a random number between 0 and 1
        random_number = np.random.random()

        # If the random number is less than r * dt,
        # termination occurs
        if random_number < probability:
            break

    # Convert number of periods into actual time
    time = periods * dt #unit conversion, whatever the unit of time

    # Save the waiting time
    waiting_times.append(time)

# Convert the list to a NumPy array
waiting_times = np.array(waiting_times)

# Plot the ECDF
sns.ecdfplot(x=waiting_times)

# Add the theoretical Exponential CDF
x = np.linspace(0, np.percentile(waiting_times, 99.5), 500)
plt.plot(x, stats.expon.cdf(x, scale=1/r))

plt.xlabel("Waiting time")
plt.ylabel("ECDF")
plt.title("Question 4: Waiting Time Distribution")

plt.show()


```
